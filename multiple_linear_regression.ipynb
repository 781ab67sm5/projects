```python
# Training set
X_train = np.array([[2104, 5, 1, 45], [1416, 3, 2, 40], [852, 2, 1, 35]])
y_train = np.array([460, 232, 178])
```

```python
# Parameter vectors
b_init = 785.1811367994083
w_init = np.array([ 0.39133535, 18.75376741, -53.36032453, -26.42131618])
print(f"w_init shape: {w_init.shape}, b_init type: {type(b_init)}")
```

```python
# Single prediction
def single_prediction(x, w, b):
    """
    Predict a house price using untuned parameters with 
        a single training example.

    Args:
        x (ndarray): Shape (n,) example with multiple features
        w (ndarray): Shape (n,) model parameters
        b (scalar):  model parameter

    Returns:
        p (scalar):  house price prediction
    """
    m = x.shape[0]
    p = 0.0
    for i in range(m):
        p_i = w[i] * x[i]
        p = p + p_i
    p = p + b
    return p
```

```python
price = single_prediction(X_train[0], w_init, b_init)
print(price)
```

```python
# Single prediction, vectorised
def single_prediction_vectorised(x, w, b):
    """
    Predict a house price using untuned parameters with 
        a single training example, now using the NumPy dot() method.

    Args:
        x (ndarray): Shape (n,) example with multiple features
        w (ndarray): Shape (n,) model parameters
        b (scalar) : model parameter

    Returns:
        p (scalar): prediction
    """
    f_wb = np.dot(x, w) + b
    return f_wb
```

```python
house_price = single_prediction_vectorised(X_train[0], w_init, b_init)
print(house_price)
```

```python
# Compute cost for multiple variable linear regression
def cost(X, y, w, b):
    """
    Compute the cost of the multiple linear regression model function.

    Args: 
        X - matrix of the training data
        y - target values (vector)
        w - weights (vector)
        b - parameters (vector)

    Returns:
        Cost for the multiple linear regression model function.
    """
    m = X.shape[0]
    cost = 0.0
    for i in range(m):
        f_wb_i = np.dot(X[i],w) + b  # For every training example...
        cost = cost + (f_wb_i - y[i]) ** 2
    cost = cost / (2 * m)
    return cost
```

```python
cost = cost(X_train, y_train, w_init, b_init)
print(cost)
```

```python
# Compute gradients
def grad(X, y, w, b):
    """
    Args:
        X - matrix of the training set
        y - vector of target values
        w - vector of parameters
        b - vector of parameters
    """
    m, n = X.shape
    dJ_dw = np.zeros(n)
    dJ_db = 0.0
    # Gradients computed for each feature, for each training example
    for i in range(m):
        f_wb_i = np.dot(X[i],w) + b  # Compute model based on example
        dJ_db = dJ_db + (f_wb_i - y[i])
        for j in range(n):
            dJ_dw[j] = dJ_dw[j] + (f_wb_i - y[i]) * X[i,j]
    dJ_dw = dJ_dw / m
    dJ_db = dJ_db / m
    return dJ_dw, dJ_db
```

```python
dJ_dw, dJ_db = grad(X_train, y_train, w_init, b_init)
print(f"dJ_dw: {dJ_dw}")
print(f"dJ_db: {dJ_db}")
```
